From the analysis, Activision appears to be one of the publishers that dominates the market. One possible reason is the popularity of franchises such as Call of Duty, which belongs mainly to the Shooter and Action genres.

Since the Call of Duty franchise has received almost yearly releases since the mid-2000s, Activision has maintained a strong and consistent market presence over time.

This suggests that long-running franchises with frequent releases can significantly influence both publisher sales and genre popularity.

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
```

```python
df["year"] = df["release_date"].dt.year
```

```python
yearly_sales = df.groupby("year")["total_sales"].sum()

yearly_sales
```

```python
yearly_sales = df.groupby("year")["total_sales"].sum().sort_index()

yearly_sales
```

```python
import matplotlib.pyplot as plt

yearly_sales.plot(figsize=(12,6))

plt.title("Global Video Game Sales Over Time")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.show()
```

Global video game sales experienced strong growth from the mid-1990s through the late 2000s, reaching a peak around 2008. This period likely reflects the expansion of major gaming consoles such as the Xbox 360, PlayStation 3, and Nintendo Wii, alongside the rise of highly successful franchises and online gaming.

After 2011, sales appear to decline gradually. However, this may also be influenced by incomplete recent data or changes in how game sales are reported in modern digital marketplaces.

# **critic_score**

```python
df[["critic_score", "total_sales"]].corr()
```

Critical scores influence commercial success, but they do not guarantee “hype.”

When analyzing the relationship between critical scores (critic_score) and overall commercial performance (total_sales), a weak positive correlation of 0.28 is observed.

Although the trend suggests that better reviews tend to be accompanied by higher revenue, this value demonstrates that a high score is not the sole driver of sales in the industry. External factors such as the marketing campaign, franchise loyalty, the game’s genre, or the release date play a potentially more decisive role in the final volume of copies sold.

# **How many games were released each year?**

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
df["year"] = df["release_date"].dt.year
games_per_year = df.groupby("year")["title"].count()

games_per_year
```

Boom, Saturation, and Maturity in the Industry’s Release Volume

A historical analysis of release volume reveals three critical stages in the video game industry. Following sustained growth throughout the 1990s, the market experienced massive expansion in the 2000s, reaching its all-time peak in 2009 with 4,360 titles released.

This phenomenon of overproduction was driven by the popularity of consoles with massive adoption (such as the Nintendo Wii and PlayStation 2) and the boom in lower-budget software development. Subsequently, the data shows a stabilization and decline in the following years. This does not imply a market contraction, but rather a transition toward commercial maturity, where the industry began to prioritize games as a service (Live Services), long-term content updates, and digital distribution focused on higher-budget titles with longer market lifespans.
